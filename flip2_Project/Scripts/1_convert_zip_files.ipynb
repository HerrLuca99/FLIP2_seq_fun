```python
# Specify the directory path
directory = '/mnt/c/Users/lherrmann/Desktop/FLIP2_Project/'

# Check if the folder exists
if not os.path.exists(directory):
    print(f"Path '{directory}' not found. Defaulting to '/mnt/uploads/'.")
    directory = '/mnt/uploads/'

# List all files in the directory
try:
    files = os.listdir(directory)
    # Filter for files that end in .gz
    gz_files = [f for f in files if f.endswith('.gz')]

    if not gz_files:
        print("No .gz files found in the directory.")
    else:
        for filename in gz_files:
            input_path = os.path.join(directory, filename)
            # Create output path by removing the .gz extension
            # (filename[:-3] removes '.gz')
            output_path = os.path.join(directory, filename[:-3])

            print(f"Extracting: {filename} ...")
            
            try:
                with gzip.open(input_path, 'rb') as f_in:
                    with open(output_path, 'wb') as f_out:
                        shutil.copyfileobj(f_in, f_out)
                print(f"  Done: {output_path}")
            except Exception as e:
                print(f"  Error extracting {filename}: {e}")

except Exception as e:
    print(f"An error occurred accessing the directory: {e}")
```
