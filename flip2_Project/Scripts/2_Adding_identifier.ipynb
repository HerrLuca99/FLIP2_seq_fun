```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
folder_path ='/home/lherrmann/projects/enzyme-design/flip2_Project/data/'
all_files = os.listdir(folder_path)
print(all_files)
# df = pd.read_csv(file_path)
```

```python
# 4. Loop through each CSV file and perform the conversion
for filename in all_files:
    # Construct the full file path
    file_path = os.path.join(folder_path, filename)
    
    # Read the CSV
    df = pd.read_csv(file_path)

    # Get 'Imine_Reductase_two_to_many.csv'
    base_name = os.path.basename(file_path)
    print(base_name)

    # Remove the extension to get 'Imine_Reductase_two_to_many'
    identifier = base_name[0] + base_name.split('_')[1][0] + base_name.split('_')[2][0]
    print(identifier)

    def add_id(x):
        return identifier+ '_'+ str(x+1)  

    df['id'] = df.index.to_series().apply(add_id)
    df['id'] = [f"{identifier}_{i+1:06d}" for i in range(len(df))]

    df = df[['id','sequence','set','validation','target']]
    
    # Define the output filename
    df.to_csv('/home/lherrmann/projects/enzyme-design/flip2_Project/test/csv_id/' + base_name[:-4] + '_id.csv', index=False)
    
print("All conversions completed.")
```
