```python
import pandas as pd
import glob
import os

# 1. Define the generic path where your CSVs live and where to save the output
# (You will replace these with your actual paths)
input_folder = "/home/lherrmann/projects/enzyme-design/flip2_Project/test/csv_id/files_index/*.csv" 
output_file = "/home/lherrmann/projects/enzyme-design/flip2_Project/test/csv_id/pickel/merged.csv"

```

```python
# 2. Use glob to get a list of all CSV files matching the pattern
csv_files = glob.glob(input_folder)

# Create an empty list to hold our dataframes
dataframe_list = []

# 3. Loop through the list of files
for file in csv_files:
    # Read the current CSV
    df = pd.read_csv(file)
    
   
    # Add it to our list
    dataframe_list.append(df)

# 4. Concatenate all dataframes in the list into one
# ignore_index=True resets the row numbers so they are continuous
combined_df = pd.concat(dataframe_list, ignore_index=True)

# 5. Save the final dataframe to a new CSV
# index=False prevents pandas from writing the row numbers as a new column
combined_df.to_csv(output_file, index=False)
```
